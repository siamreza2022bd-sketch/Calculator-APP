```python
import random
import math
import pandas as pd
import folium

random.seed(42)

# ==========================================================
# MANHATTAN DISTANCE
# ==========================================================
def manhattan_distance(lat1, lon1, lat2, lon2):
    lat_km = abs(lat2 - lat1) * 111
    lon_km = abs(lon2 - lon1) * 111 * math.cos(math.radians((lat1 + lat2) / 2))
    return lat_km + lon_km


# ==========================================================
# DATA GENERATION
# ==========================================================
def generate_election_data():

    voting_centers = []
    for i in range(1, 101):
        center = {
            'center_id': f'VC-{i:03d}',
            'name': f'Voting Center {i}',
            'location': (random.uniform(23.70, 23.90),
                         random.uniform(90.30, 90.50)),
            'capacity': 1500
        }
        voting_centers.append(center)

    count_centers = []
    for i in range(1, 11):
        count_center = {
            'count_id': f'CC-{i:02d}',
            'name': f'Count Center {i}',
            'location': (random.uniform(23.70, 23.90),
                         random.uniform(90.30, 90.50))
        }
        count_centers.append(count_center)

    return voting_centers, count_centers


# ==========================================================
# MAIN
# ==========================================================
voting_centers, count_centers = generate_election_data()

# Collect all coordinates
all_locations = [vc['location'] for vc in voting_centers] + \
                [cc['location'] for cc in count_centers]

latitudes = [loc[0] for loc in all_locations]
longitudes = [loc[1] for loc in all_locations]

# Compute bounding box
min_lat, max_lat = min(latitudes), max(latitudes)
min_lon, max_lon = min(longitudes), max(longitudes)

# Add small padding
padding = 0.01
min_lat -= padding
max_lat += padding
min_lon -= padding
max_lon += padding

# ==========================================================
# CREATE MAP
# ==========================================================
city_map = folium.Map(location=[23.80, 90.40], zoom_start=12)

# Add Border Polygon (City Boundary)
boundary = [
    [min_lat, min_lon],
    [min_lat, max_lon],
    [max_lat, max_lon],
    [max_lat, min_lon]
]

folium.Polygon(
    locations=boundary,
    color="green",
    weight=3,
    fill=True,
    fill_opacity=0.05,
    popup="Election Coverage Area"
).add_to(city_map)

# Voting Centers (Blue)
for vc in voting_centers:
    folium.CircleMarker(
        location=vc['location'],
        radius=4,
        color='blue',
        fill=True,
        fill_color='blue',
        popup=f"{vc['center_id']}"
    ).add_to(city_map)

# Count Centers (Red)
for cc in count_centers:
    folium.Marker(
        location=cc['location'],
        icon=folium.Icon(color='red'),
        popup=f"{cc['count_id']}"
    ).add_to(city_map)

city_map
```

# Search Voter Information and Voting Center Information **[Milestone-1]**

```python
# ==========================================================
# MILESTONE 1
# Voter Search + Voting Center Load Analyzer
# ==========================================================

# -------- Sample Data --------
voters = [
    {"voter_id": "VOTER-000120", "name": "Rahim", "voting_center": "VC-010"},
    {"voter_id": "VOTER-000121", "name": "Karim", "voting_center": "VC-010"},
    {"voter_id": "VOTER-000122", "name": "Siam", "voting_center": "VC-011"},
]

voting_centers = [
    {"center_id": "VC-010", "name": "Dhaka College Center", "location": "23.7275, 90.3854"},
    {"center_id": "VC-011", "name": "City School Center", "location": "23.8103, 90.4125"},
]


# -------- Linear Search Voter --------
def search_voter(voter_id, voters, voting_centers):

    for voter in voters:   # Linear Search O(n)
        if voter['voter_id'] == voter_id:

            center_id = voter['voting_center']

            # find center info
            for vc in voting_centers:
                if vc['center_id'] == center_id:

                    print("\n===== VOTER INFORMATION =====")
                    print("Voter ID:", voter['voter_id'])
                    print("Name:", voter['name'])
                    print("Voting Center ID:", vc['center_id'])
                    print("Center Name:", vc['name'])
                    print("GPS Location:", vc['location'])
                    return

    print("Voter not found")


# -------- Voting Center Load Analyzer --------
def search_center(center_id, voters, voting_centers):

    center = None
    for vc in voting_centers:
        if vc['center_id'] == center_id:
            center = vc
            break

    if center is None:
        print("Center not found")
        return

    assigned_voters = []

    for voter in voters:   # Linear Search
        if voter['voting_center'] == center_id:
            assigned_voters.append(voter)

    print("\n===== CENTER INFORMATION =====")
    print("Center ID:", center['center_id'])
    print("Center Name:", center['name'])
    print("Location:", center['location'])
    print("Total Assigned Voters:", len(assigned_voters))


# -------- Example Search --------
search_voter("VOTER-000120", voters, voting_centers)
search_center("VC-010", voters, voting_centers)
```

# Election Day

On election day, we will assume a total voter turnout of 67.5%. To ensure a realistic simulation, the number of votes cast at each specific voting center will vary and be generated randomly.

This means that while the overall city-wide turnout hits the 2,02,500 mark (out of 300,000 voters), each of the 100 voting centers will handle a unique and randomized volume of ballot papers

```python
import random

# -------- Sample Data --------
voters = [
    {"voter_id": "V1", "name": "Rahim", "voting_center": "VC-010"},
    {"voter_id": "V2", "name": "Karim", "voting_center": "VC-010"},
    {"voter_id": "V3", "name": "Siam", "voting_center": "VC-011"},
    {"voter_id": "V4", "name": "Rita", "voting_center": "VC-011"},
    {"voter_id": "V5", "name": "Hasan", "voting_center": "VC-012"},
    {"voter_id": "V6", "name": "Nila", "voting_center": "VC-012"},
]

candidates = [
    {"candidate_id": "C1", "name": "Candidate A"},
    {"candidate_id": "C2", "name": "Candidate B"},
    {"candidate_id": "C3", "name": "Candidate C"},
]

voting_centers = [
    {"center_id": "VC-010", "name": "Dhaka College", "location": "23.72, 90.38"},
    {"center_id": "VC-011", "name": "City School", "location": "23.81, 90.41"},
    {"center_id": "VC-012", "name": "Govt School", "location": "23.75, 90.39"},
]


# -------- Simulation Function --------
def simulate_election_day(voters, candidates, voting_centers):
    total_voters = len(voters)
    target_turnout = int(total_voters * 0.675)

    # Randomly pick voters
    active_voters = random.sample(voters, target_turnout)

    # Initialize results
    election_results = {
        vc['center_id']: {
            'total_center_votes': 0,
            **{c['candidate_id']: 0 for c in candidates}
        } for vc in voting_centers
    }

    # Simulate voting
    for voter in active_voters:
        center_id = voter['voting_center']
        voted_candidate = random.choice(candidates)
        candidate_id = voted_candidate['candidate_id']

        election_results[center_id][candidate_id] += 1
        election_results[center_id]['total_center_votes'] += 1

    return election_results, target_turnout


# -------- Execution --------
results, actual_total = simulate_election_day(voters, candidates, voting_centers)

# -------- Output --------
print("="*70)
print("             DHAKA VOTE ENGINEERING SYSTEM: FINAL TALLY")
print("="*70)
print(f"Total Voter List        : {len(voters)}")
print(f"Vote Turnout            : {actual_total}")
print(f"Vote Casting Percentage : {(actual_total/len(voters))*100:.2f}%")
print("-" * 70)

# -------- Center-wise Results --------
for center_id, data in results.items():
    print(f"\nCenter: {center_id}")
    print(f"Total Votes: {data['total_center_votes']}")

    for cid, votes in data.items():
        if cid != 'total_center_votes':
            print(f"  {cid}: {votes}")
```

# Ballote Paper Transfer **[Milestone-2]**

```python
## ==========================================================
# MILESTONE 2
# Ballot Transport System
# Manhattan Distance + Dijkstra
# ==========================================================

import heapq

# -------- Manhattan Distance --------
def manhattan_distance(lat1, lon1, lat2, lon2):
    return abs(lat1 - lat2) + abs(lon1 - lon2)


# -------- Sample Data --------
voting_centers = [
    {"center_id": "VC-010", "location": (23.72, 90.38)},
    {"center_id": "VC-011", "location": (23.81, 90.41)},
    {"center_id": "VC-012", "location": (23.75, 90.39)},
]

count_centers = [
    {"count_id": "CC-01", "location": (23.70, 90.40)},
    {"count_id": "CC-02", "location": (23.82, 90.42)},
]


# -------- Dijkstra Algorithm --------
def dijkstra(graph, start, end):

    queue = [(0, start, [])]
    visited = set()

    while queue:
        (cost, node, path) = heapq.heappop(queue)

        if node in visited:
            continue

        path = path + [node]
        visited.add(node)

        if node == end:
            return cost, path

        for next_node, weight in graph[node]:
            heapq.heappush(queue, (cost + weight, next_node, path))

    return float("inf"), []


# -------- Build Graph --------
def build_graph(voting_centers, count_centers):

    nodes = {}
    graph = {}

    # Collect all nodes
    for vc in voting_centers:
        nodes[vc['center_id']] = vc['location']

    for cc in count_centers:
        nodes[cc['count_id']] = cc['location']

    # Fully connected graph
    for node1 in nodes:
        graph[node1] = []

        for node2 in nodes:
            if node1 != node2:
                lat1, lon1 = nodes[node1]
                lat2, lon2 = nodes[node2]

                dist = manhattan_distance(lat1, lon1, lat2, lon2)

                graph[node1].append((node2, dist))

    return graph


# -------- Build Graph --------
graph = build_graph(voting_centers, count_centers)


# -------- Assign Voting Centers to Count Centers --------
assignments = {}
count_capacity = {cc['count_id']: 0 for cc in count_centers}

for vc in voting_centers:

    vc_id = vc['center_id']
    vc_lat, vc_lon = vc['location']

    distances = []

    for cc in count_centers:
        cc_lat, cc_lon = cc['location']

        d = manhattan_distance(vc_lat, vc_lon, cc_lat, cc_lon)
        distances.append((d, cc['count_id']))

    distances.sort()

    for d, cc_id in distances:

        if count_capacity[cc_id] < 10:
            count_capacity[cc_id] += 1

            cost, path = dijkstra(graph, vc_id, cc_id)

            assignments[vc_id] = {
                "count_center": cc_id,
                "distance": cost,
                "path": path
            }

            break


# -------- Output --------
print("Ballot Transfer Assignment Completed\n")

for vc_id, info in assignments.items():
    print(f"Voting Center: {vc_id}")
    print(f"  Assigned Count Center: {info['count_center']}")
    print(f"  Shortest Distance: {info['distance']:.4f}")
    print(f"  Path: {' -> '.join(info['path'])}")
    print("-" * 50)
```

```python
# ==========================================================
# MILESTONE 3
# Final Vote Counting
# ==========================================================

# -------- Sample Candidates (FIXED) --------
candidates = [
    {"candidate_id": "C1", "name": "Candidate A", "marka": "Boat", "logo": "🚤"},
    {"candidate_id": "C2", "name": "Candidate B", "marka": "Tiger", "logo": "🐯"},
    {"candidate_id": "C3", "name": "Candidate C", "marka": "Eagle", "logo": "🦅"},
]

# -------- Sample Results (from simulation) --------
results = {
    "VC-010": {"total_center_votes": 3, "C1": 1, "C2": 1, "C3": 1},
    "VC-011": {"total_center_votes": 2, "C1": 1, "C2": 1, "C3": 0},
    "VC-012": {"total_center_votes": 2, "C1": 0, "C2": 1, "C3": 1},
}

# -------- Aggregate Votes --------
total_votes = {c['candidate_id']: 0 for c in candidates}

for center in results:
    for c in candidates:
        cid = c['candidate_id']
        total_votes[cid] += results[center].get(cid, 0)   # safe access


# -------- Convert to List --------
candidate_results = []

for c in candidates:
    cid = c['candidate_id']

    candidate_results.append({
        "candidate_id": cid,
        "name": c['name'],
        "marka": c['marka'],
        "logo": c['logo'],
        "votes": total_votes[cid]
    })


# -------- Sort (Descending for better display) --------
candidate_results.sort(key=lambda x: x["votes"], reverse=True)


# -------- Display Table --------
print("\n===== FINAL RESULT =====\n")

for r in candidate_results:
    print(f"{r['logo']} {r['name']} ({r['marka']})  ->  {r['votes']} votes")


# -------- Winner --------
winner = candidate_results[0]   # already sorted descending

print("\n==============================")
print("ELECTION WINNER")
print("==============================")

print(f"{winner['logo']} {winner['name']} with {winner['votes']} votes")
```
